### Reference

- creating a copy of a list DOES NOT create a new list. It points to the same place in memory.
- Google: Pass by value vs Pass by Reference.
- Changing one referenced list, will also change the other.

```python
og_list = [1, 2, 3, 4, 5]

new_list = og_list

print('OG', og_list)
print('New', new_list)

new_list.append(6)

print('OG', og_list)
print('New', new_list)
```

```python
# Deep copy - a completely separate copy of a Non-primitive data type, not just a reference to the same place in memory
# Shallow copy - a semi separate copy of a Non-primitive data type. 1st dimension is separate places in memory

second_copy = og_list.copy()

print(second_copy)
print(og_list)

og_list.append(7)

print(second_copy)
print(og_list)
```

```python
# reverse(), extend(), sort()
num_list = [1, 2, 3, 4, 5]
print(num_list)

num_list.reverse()
print(num_list)

num_list.extend([6, 7, 8, 9])
print(num_list)

num_list.sort()
print(num_list)
```

```python
# Membership operater

print(1 in num_list)
print(1 not in num_list)
```

```python
# Create a function. take in 2 lists of years, reverse chronological order, add them together
# Create reusable codeblock, can be reuse as many times as needed
# CAN take input, AND can provide output

# Define a function called 'reoder_profit_years'
# Takes 2 parameters (years1, years2)
def reoder_profit_years(year1, year2):

    # Reversed the order of years for both lists
    years1.reverse()
    years2.reverse()

    # Added lists together
    years1.extend(years2)
    
    # Return the newly extended years
    return years1
```
